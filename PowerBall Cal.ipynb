```python
#predicting the next value in a column [PowerBall]

# maximum allowed value
n = 26  # change as needed

# prepare data
y = dataframe["PowerBall"].dropna().to_numpy()
X = np.arange(len(y)).reshape(-1, 1)

# train model
model = LinearRegression()
model.fit(X, y)

# predict next value
raw_prediction = model.predict([[len(y)]])[0]

# enforce integer + bounds
predicted_int = int(round(raw_prediction))
predicted_int = max(2, min(predicted_int, n))

print("Predicted next value:", predicted_int)
```
